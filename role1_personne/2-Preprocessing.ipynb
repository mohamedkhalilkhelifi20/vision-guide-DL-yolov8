## 1. Importer les bibliothèques

```python
import pandas as pd
import os
print("ok")
```

## 2. Recharger les données

```python
data_path = "/kaggle/input/datasets/aibloy/fairface/FairFace"

train_df = pd.read_csv(f"{data_path}/train_labels.csv")
val_df   = pd.read_csv(f"{data_path}/val_labels.csv")

# Ajouter chemin complet
train_df["chemin"] = data_path + "/" + train_df["file"]
val_df["chemin"]   = data_path + "/" + val_df["file"]

print(f"Train : {len(train_df)} images")
print(f"Val   : {len(val_df)} images")
print(train_df.head(3))
```

## 3. Mapping age to classe

```python
def age_to_class(age: str) -> str:
    if age in ["0-2", "3-9", "10-19"]:
        return "enfant"
    elif age == "more than 70" or int(age.split("-")[0]) >= 50:
        return "personne_agee"
    else:
        return "adulte"

train_df["classe"] = train_df["age"].apply(age_to_class)
val_df["classe"]   = val_df["age"].apply(age_to_class)

print("Train :")
print(train_df["classe"].value_counts())
print("\nVal :")
print(val_df["classe"].value_counts())
```

## 4. Vérification

```python
print("NaN train :", train_df["classe"].isna().sum())
print("NaN val   :", val_df["classe"].isna().sum())

missing = train_df[~train_df["chemin"].apply(os.path.exists)]
print(f"Images manquantes : {len(missing)}")
```

## 5. Sauvegarde

```python
train_df[["chemin", "classe"]].to_csv("/kaggle/working/train.csv", index=False)
val_df[["chemin", "classe"]].to_csv("/kaggle/working/val.csv", index=False)

print(f"Train sauvegardé : {len(train_df)} images")
print(f"Val   sauvegardé : {len(val_df)} images")
```
